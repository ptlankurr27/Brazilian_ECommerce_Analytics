```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
customers = pd.read_csv(r'D:\Ecommerce_Analytics_Project\Data\Cleaned\customers_clean.csv')
orders = pd.read_csv(r'D:\Ecommerce_Analytics_Project\Data\Cleaned\orders_clean.csv')
products = pd.read_csv(r'D:\Ecommerce_Analytics_Project\Data\Cleaned\products_clean.csv')
order_items = pd.read_csv(r'D:\Ecommerce_Analytics_Project\Data\Cleaned\order_items_clean.csv')
payments = pd.read_csv(r'D:\Ecommerce_Analytics_Project\Data\Cleaned\payments_clean.csv')
reviews = pd.read_csv(r'D:\Ecommerce_Analytics_Project\Data\Cleaned\reviews_clean.csv')
sellers = pd.read_csv(r'D:\Ecommerce_Analytics_Project\Data\Cleaned\sellers_clean.csv')
geolocation = pd.read_csv(r'D:\Ecommerce_Analytics_Project\Data\Cleaned\geolocation_clean.csv')
```

Revenue Trend Analysis

```python
df = pd.merge(
    orders, payments,
    on = 'order_id',
    how= 'inner'
              )

df.head()
```

```python
monthly_revenue = (
    df.groupby(['order_year', 'order_month'])['payment_value']
      .sum()
      .reset_index()
      .sort_values(by=['order_year', 'order_month'])
)
monthly_revenue.head(10)
```

```python
plt.figure(figsize=(12, 6))

sns.lineplot(
    data=monthly_revenue,
    x="order_month",
    y="payment_value",
    hue='order_year',   # hue tells Seaborn to split the data into separate colored lines based on order_year
    marker="o"
)

plt.title("Monthly Revenue Trend", fontsize=16)
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.grid(True)                 # Adds horizontal and vertical grid lines.
plt.tight_layout()             # Automatically adjusts spacing so labels and titles don't get cut off.

plt.show()
```

Orders Trend


```python
monthly_orders = (
    orders.groupby(['order_year', 'order_month'])
          .size()
          .reset_index(name='total_orders')
          .sort_values(by=['order_year', 'order_month'])
)
monthly_orders.head(10)
```

```python
plt.figure(figsize=(12,6))

sns.lineplot(
    data=monthly_orders,
    x='order_month',
    y='total_orders',
    hue='order_year',
    marker='o'
)

plt.title('Monthly Orders Trend')
plt.xlabel('Year-Month')
plt.ylabel('Total Orders')
plt.grid(True)
plt.tight_layout()

plt.show()
```

Revenue by Product Category

```python
df1 = pd.merge(
    order_items, products,
    on = 'product_id',
    how= 'inner'
              )

df1.head()
```

```python
product_revenue = (
    df1.groupby(['product_category_name'])['price']
       .sum()
       .reset_index()
       .sort_values(by='price', ascending=False)
)

top10 = product_revenue.head(10)
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=top10,
    x='product_category_name',
    y = 'price'
)

plt.title("Revenue by Product Category")
plt.xlabel("Product category")
plt.ylabel("Revenue")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
```

Top 10 Sellers by Revenue

```python
df3 = pd.merge(
    order_items, sellers,
    on='seller_id',
    how='inner'
)
df3.head()
```

```python
seller_revenue = (
    df3.groupby(['seller_id'])['price']
    .sum()
    .reset_index()
    .sort_values(by='price', ascending=False)
)
seller10 = seller_revenue.head(10)
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    data=seller10,
    x='price',
    y='seller_id'
)

plt.title("Top 10 Sellers by Revenue")
plt.xlabel("Revenue")
plt.ylabel("Top Seller")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

Revenue by State

```python
dfa = pd.merge(
    customers, orders,
    on= 'customer_id', 
    how='inner'
)

dfb = pd.merge(
    dfa, payments,
    on= 'order_id',
    how='inner'
)

dfb.head()
```

```python
state_revenue = (
    dfb.groupby(['customer_state'])['payment_value']
    .sum()
    .reset_index()
    .sort_values(by='payment_value')
)
state_revenue.head(10)
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    data=state_revenue,
    x = 'customer_state',
    y = 'payment_value'
)

plt.title("Revenue by State")
plt.xlabel("State")
plt.ylabel("Revenue")
plt.tight_layout()
plt.show()
```

Payment Method Distribution

```python
payment_method = (
    payments.groupby(['payment_type'])['order_id']
    .count()
    .reset_index()
    .sort_values(by='order_id', ascending=False)
    )
payment_method
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    data=payment_method,
    x='payment_type',
    y='order_id'

)
plt.title("Payment Method Distribution")
plt.xlabel("Payment Methods")
plt.ylabel("Number of Orders")
plt.tight_layout()
plt.show()
```

```python
# Using countplot()
plt.figure(figsize=(12,6))

sns.countplot(
    data=payments,
    x='payment_type',
    order=payments['payment_type'].value_counts().index
)

plt.title("Payment Method Distribution")
plt.xlabel("Payment Method")
plt.ylabel("Number of Orders")
plt.tight_layout()
plt.show()
```

Rating Distribution

```python
rate_dist = (
    reviews.groupby(['review_score'])['order_id']
    .count()
    .reset_index()
    .sort_values(by='order_id', ascending=False)
)
rate_dist
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    data=rate_dist,
    x='order_id',
    y='review_score'
)

plt.title("Rating Distribution")
plt.ylabel("Rating")
plt.xlabel("Order with Rating")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.countplot(
    data=reviews,
    y='review_score',
    order=[1, 2, 3, 4, 5]
)

plt.title("Rating Distribution")
plt.ylabel("Rating")
plt.xlabel("Number of Reviews")
plt.tight_layout()
plt.show()
```

Delivery Time Distribution

```python
orders['order_purchase_timestamp'] = pd.to_datetime(
    orders['order_purchase_timestamp']
)

orders['order_delivered_customer_date'] = pd.to_datetime(
    orders['order_delivered_customer_date']
)
```

```python
# Create a New Column
orders['delivery_days'] = (
    orders['order_delivered_customer_date'] -
    orders['order_purchase_timestamp']
).dt.days
```

```python
plt.figure(figsize=(12,6))

sns.histplot(
    data=orders,
    x='delivery_days',
    bins=30              # Histogram divides the data into intervals (bins).
)

plt.title("Delivery Time Distribution")
plt.xlabel("Delivery Days")
plt.ylabel("Number of Orders")
plt.tight_layout()
plt.show()
```

Correlation Heatmap

```python
corr_df = products[
    [
        'product_weight_g',
        'product_length_cm',
        'product_height_cm',
        'product_width_cm'
    ]
]
```

```python
corr = corr_df.corr()   # calculates how strongly two numeric columns are related.
```

```python
plt.figure(figsize=(10,8))

sns.heatmap(
    corr,
    annot=True,             # Displays the correlation values inside each cell.
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```
